```python
import os

import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd

# Display all columns when looking at dataframes
pd.set_option('display.max_columns', None)
```

```python
# Reading in earthquake data
url = ("https://raw.githubusercontent.com/carmengg/eds220-hwk-1/main/data/earthquake_data.csv")
earthquakes = pd.read_csv(url)

# Reading in countries data
fp = os.path.join('data_countries', 'ne_110m_admin_0_countries.shp')
countries = gpd.read_file(fp)
```

### 2. Recreate the plot

Filtering to earthquakes with a magnitude 6.5 or greater

```python
# Filtering to 6.5+ magnitude

filter_earthquakes = earthquakes[earthquakes['mag'] >= 6.5]
```

Creting a `geodataframe` from a CSV

```python
# Create geodataframe from CSV

# Create points from latitude and longitude
points = gpd.points_from_xy(filter_earthquakes['longitude'],
                   filter_earthquakes['latitude'])

# Create geodataframe
earthquake_mags = gpd.GeoDataFrame(filter_earthquakes, # Data
                                geometry=points,
                                crs='EPSG:4326')
```

Making sure the CRSs match

```python
# Check CRSs match

earthquake_mags.crs == countries.crs
```

Plotting earthquake magnitudes

```python
fig, ax = plt.subplots()

# Plotting world map
countries.plot(ax=ax,
               color = 'lightblue')

# Plotting earthquale magnitudes
earthquake_mags.plot(ax=ax,
                     alpha = 0.5,
                     c='mag',
                     cmap='RdPu_r'
                     )
# Update axes labels
ax.set_title('Earthquakes of Magnitude at least 6.5 in 2014')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude');
```
